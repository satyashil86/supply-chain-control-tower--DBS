```python
%sql
-- Step 3 — Grant Data Engineer permissions
-- SC_DATA_ENGINEER should be able to work with all three data layers.

-- Catalog access
GRANT USE CATALOG
ON CATALOG workspace
TO `SC_DATA_ENGINEER`;

-- BRONZE
GRANT USE SCHEMA
ON SCHEMA workspace.bronze
TO `SC_DATA_ENGINEER`;

GRANT SELECT, MODIFY
ON SCHEMA workspace.bronze
TO `SC_DATA_ENGINEER`;

-- SILVER
GRANT USE SCHEMA
ON SCHEMA workspace.silver
TO `SC_DATA_ENGINEER`;

GRANT SELECT, MODIFY
ON SCHEMA workspace.silver
TO `SC_DATA_ENGINEER`;

-- GOLD
GRANT USE SCHEMA
ON SCHEMA workspace.gold
TO `SC_DATA_ENGINEER`;

GRANT SELECT, MODIFY
ON SCHEMA workspace.gold
TO `SC_DATA_ENGINEER`;

```

```python
%sql
-- Step 4 — Grant Planner permissions
-- The planner should read Silver and Gold, but not modify them

GRANT USE CATALOG
ON CATALOG workspace
TO `SC_PLANNER`;

-- SILVER
GRANT USE SCHEMA
ON SCHEMA workspace.silver
TO `SC_PLANNER`;

GRANT SELECT
ON SCHEMA workspace.silver
TO `SC_PLANNER`;

-- GOLD
GRANT USE SCHEMA
ON SCHEMA workspace.gold
TO `SC_PLANNER`;

GRANT SELECT
ON SCHEMA workspace.gold
TO `SC_PLANNER`;
```

```python
%sql
-- Step 5 — Grant Viewer permissions
-- The viewer should only consume the final Gold outputs.

GRANT USE CATALOG
ON CATALOG workspace
TO `SC_VIEWER`;

GRANT USE SCHEMA
ON SCHEMA workspace.gold
TO `SC_VIEWER`;

GRANT SELECT
ON SCHEMA workspace.gold
TO `SC_VIEWER`;
```

```python
%sql
-- Step 6 — Audit permissions
-- Since we have an audit schema, give the Data Engineer access to it

GRANT USE CATALOG
ON CATALOG workspace
TO `SC_DATA_ENGINEER`;

GRANT USE SCHEMA
ON SCHEMA workspace.audit
TO `SC_DATA_ENGINEER`;

GRANT SELECT
ON SCHEMA workspace.audit
TO `SC_DATA_ENGINEER`;
```
