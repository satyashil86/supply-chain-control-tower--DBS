```python
from pyspark.sql import functions as F

#Source file
path="/Volumes/workspace/supply_chain/landing/SAP/MARA.csv"
# Read csv
df_mara = (
    spark.read
    .option("header","true")
    .option("inferschema","true")
    .csv(path)
)
#check data
display(df_mara)
```

```python
df_mara.printSchema()
```

```python
# Write MARA to Bronze
df_mara.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.bronze.mara")
```

```python
# Step 3D — Check that it is really a Delta table
spark.sql("DESCRIBE DETAIL workspace.bronze.mara")
```

```python
from pyspark.sql import functions as F

# ============================================================
# SAP TABLES
# ============================================================

sap_tables = {
    "MARC": "marc",
    "MARD": "mard",
    "T001W": "t001w",
    "LFA1": "lfa1",
    "VBAK": "vbak",
    "VBAP": "vbap",
    "EKKO": "ekko",
    "EKPO": "ekpo",
    "EKET": "eket"
}

for file_name, table_name in sap_tables.items():

    path = f"/Volumes/workspace/supply_chain/landing/SAP/{file_name}.csv"

    df = (
        spark.read
        .option("header", "true")
        .option("inferSchema", "true")
        .csv(path)
    )

    df.write \
        .format("delta") \
        .mode("overwrite") \
        .saveAsTable(f"workspace.bronze.{table_name}")

    print(f"Loaded SAP: {file_name}.csv -> bronze.{table_name}")


# ============================================================
# TMS TABLES
# ============================================================

tms_tables = {
    "shipment_header": "shipment_header",
    "tracking_event": "tracking_event",
    "carrier_master": "carrier_master"
}

for file_name, table_name in tms_tables.items():

    path = f"/Volumes/workspace/supply_chain/landing/TMS/{file_name}.csv"

    df = (
        spark.read
        .option("header", "true")
        .option("inferSchema", "true")
        .csv(path)
    )

    df.write \
        .format("delta") \
        .mode("overwrite") \
        .saveAsTable(f"workspace.bronze.{table_name}")

    print(f"Loaded TMS: {file_name}.csv -> bronze.{table_name}")


# ============================================================
# WMS TABLES
# ============================================================

wms_tables = {
    "wms_inventory": "wms_inventory",
    "pick_task": "pick_task"
}

for file_name, table_name in wms_tables.items():

    path = f"/Volumes/workspace/supply_chain/landing/WMS/{file_name}.csv"

    df = (
        spark.read
        .option("header", "true")
        .option("inferSchema", "true")
        .csv(path)
    )

    df.write \
        .format("delta") \
        .mode("overwrite") \
        .saveAsTable(f"workspace.bronze.{table_name}")

    print(f"Loaded WMS: {file_name}.csv -> bronze.{table_name}")

print("==========================================")
print("All 14 remaining Bronze tables loaded.")
print("==========================================")
```
